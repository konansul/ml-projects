```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from sklearn.cluster import KMeans

%config InlineBackend.figure_format = 'retina'
```

```python
image = mpimg.imread('/Users/konansul/Desktop/course-ml/DATA/palm_trees.jpg')
```

```python
image.shape
```

```python
plt.imshow(image)
```

```python
(h, w, c) = image.shape
```

```python
image_2d = image.reshape(h * w, c)
```

```python
image_2d
```

```python
image_2d.shape
```

```python
model = KMeans(n_clusters = 6)
```

```python
labels = model.fit_predict(image_2d)
```

```python
labels
```

```python
rgb = model.cluster_centers_.round(0).astype(int)
```

```python
rgb
```

```python
rgb[labels]
```

```python
image_qunat = np.reshape(rgb[labels], (h, w, c))
```

```python
plt.imshow(image)
```

```python
plt.imshow(image_qunat)
```
