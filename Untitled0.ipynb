```python
import itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
from sklearn.preprocessing import StandardScaler
%config InlineBackend.figure_format = 'retina'
```

```python
x = np.concatenate((np.linspace(-3, 0, 500), np.linspace(0, 3, 500)))
y = np.ones(1000)
w = np.ones(1000)
M = y * x * w
plt.plot(x * w, x < 0, label = 'Zero-one loss')
plt.plot(M, np.log2(1 + np.exp(-M)), label = 'Logistic loss')
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/konansul/ml-projects/main/02%20%E2%80%93%20Logistic%20Regression/data/iris.csv')
```

```python
data
```

```python
data.info()
```

```python
sns.pairplot(data, hue = 'species')
```

```python
y = data['species'].map({'setosa': 1, 'versicolor': 2, 'virginica': 3})
```

```python
X = data[['sepal_length', 'sepal_width']]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 0)
```

```python
scaler = StandardScaler()
```

```python
scaled_x_train = scaler.fit_transform(X_train)
scaled_x_test = scaler.transform(X_test)
```

```python
logistic_model = LogisticRegression()
```

```python
logistic_model.fit(scaled_x_train, y_train)
```

```python
logistic_pred = logistic_model.predict(scaled_x_test)
```

```python
print(classification_report(y_test, logistic_pred))
```

```python
svm_model = SVC(kernel = 'linear')
```

```python
svm_model.fit(scaled_x_train, y_train)
```

```python
svm_pred = svm_model.predict(scaled_x_test)
```

```python
print(classification_report(y_test, svm_pred))
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
gs = gridspec.GridSpec(1, 2)
fig = plt.figure(figsize = (12, 6))
labels = ['Logistic Regression', 'SVM']

for clf, lab, grd in zip([logistic_model, svm_model], labels, itertools.product([0, 1], repeat = 2)):
    ax = plt.subplot(gs[grd[0], grd[1]])
    fig = plot_decision_regions(X = np.array(scaled_x_train), y = np.array(y_train), clf = clf, legend = 2)
    plt.title(lab)
plt.show()
```

```python
log_model = LogisticRegression(solver = 'saga', max_iter = 5000)
```

```python
penalty = ['l1', 'l2', 'elasticnet']
l1_ratio = np.linspace(0, 1, 20)
C = np.logspace(0, 10, 20)
```

```python
parameter_grid = {'penalty': penalty, 'l1_ratio' : l1_ratio, 'C': C}
```

```python
log_grid = GridSearchCV(log_model, param_grid = parameter_grid, n_jobs = -1)
```

```python
log_grid.fit(scaled_x_train, y_train)
```

```python
log_grid.best_params_
```

```python
print(classification_report(y_test, log_grid.predict(scaled_x_test)))
```

```python
ConfusionMatrixDisplay.from_estimator(log_grid, scaled_x_test, y_test, cmap = 'Blues')
```
